```python
import requests
from bs4 import BeautifulSoup
import json

# Define the URL of the webpage
url = 'https://worldathletics.org/results/olympic-games/2024/the-xxxiii-olympic-games-7153115/men/5000-metres/final/result'

# Send a GET request to the webpage
response = requests.get(url)

# Check if the request was successful
if response.status_code == 200:
    # Parse the HTML content of the webpage
    soup = BeautifulSoup(response.content, 'html.parser')

    # Find all tables on the page
    tables = soup.find_all('table', class_='records-table clickable')

    # Initialize a list to store the results
    results = []

    # Process each table
    for table in tables:
        # Find all rows in the table body
        rows = table.find('tbody').find_all('tr')

        for row in rows:
            # Find the 'POS' column (Position)
            pos_td = row.find('td', {'data-th': 'POS'})
            pos = pos_td.get_text(strip=True) if pos_td else None

            # Find the 'MARK' column (Time)
            mark_td = row.find('td', {'data-th': 'MARK'})
            mark = mark_td.get_text(strip=True) if mark_td else None

            # Find the 'ATHLETE' column (Name)
            athlete_td = row.find('td', {'data-th': 'athlete'})
            name = athlete_td.get_text(strip=True) if athlete_td else None

            # Store the extracted data in a dictionary
            if pos or mark or name:
                result = {
                    'Position': pos,
                    'Name': name,
                    'Time': mark
                }
                results.append(result)
else:
    print(f"Failed to retrieve the webpage. Status code: {response.status_code}")

# Save the results to a JSON file 
with open('../data/final_results.json', 'w') as json_file:
    json.dump(results, json_file, indent=4)
```
